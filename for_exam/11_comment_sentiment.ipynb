```python
import requests, pandas as pd, matplotlib.pyplot as plt
from bs4 import BeautifulSoup
from textblob import TextBlob

try:
    r = requests.get('https://www.sahyadri.edu.in/', headers={'User-Agent': 'Mozilla/5.0'}, timeout=15); r.raise_for_status()
    comments = [c.get_text(' ', strip=True) for c in BeautifulSoup(r.text, 'html.parser').select('p.comment-text')]
except Exception as e:
    print('Could not scrape the webpage:', e); comments = []

if not comments:
    print('No comments found - using sample comments.')
    comments = ['I love this post! Very informative.', 'This article is terrible and full of mistakes.', 'Pretty good, but it needs more examples.', 'I completely disagree with the author.', 'Excellent explanation, thanks for sharing!']
```

```python
df = pd.DataFrame({'Comment': comments})
df['Sentiment'] = df['Comment'].apply(lambda t: 'Positive' if TextBlob(t).sentiment.polarity > 0 else ('Negative' if TextBlob(t).sentiment.polarity < 0 else 'Neutral'))
print(df.to_string(index=False))

counts = df['Sentiment'].value_counts().reindex(['Positive', 'Negative', 'Neutral'], fill_value=0)
print('\nSentiment Counts:\n', counts)
```

```python
counts.plot(kind='bar', rot=0, color=['green', 'red', 'gray'], title='Sentiment Distribution')
plt.xlabel('Sentiment'); plt.ylabel('Number of Comments'); plt.tight_layout(); plt.show()

counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, title='Sentiment Distribution')
plt.ylabel(''); plt.tight_layout(); plt.show()
```
