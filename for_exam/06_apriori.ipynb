```python
import pandas as pd, os
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
if os.path.basename(os.getcwd()) == 'for_exam': os.chdir('..')

transactions = [['milk','bread','butter'],['bread','butter'],['milk','bread'],['milk','butter'],['bread','eggs'],['milk','bread','eggs'],['bread','butter'],['milk','bread','butter'],['milk','eggs'],['bread','butter','eggs'],['milk','bread'],['bread','butter']]
```

```python
te = TransactionEncoder()
df = pd.DataFrame(te.fit(transactions).transform(transactions), columns=te.columns_)
df.head()
```

```python
frequent = apriori(df, min_support=0.4, use_colnames=True)
print('Frequent Itemsets:\n', frequent)

rules = association_rules(frequent, metric='confidence', min_threshold=0.6)
rules[['antecedents','consequents','support','confidence','lift']]
```
