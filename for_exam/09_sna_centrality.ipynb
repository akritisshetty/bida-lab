```python
import pandas as pd, networkx as nx, matplotlib.pyplot as plt, os
if os.path.basename(os.getcwd()) == 'for_exam': os.chdir('..')

df = pd.read_csv('data_exp9.txt', sep=r'\s+', header=None, names=['user_1', 'user_2']).dropna()
G = nx.from_pandas_edgelist(df, source='user_1', target='user_2')
print('Users:', G.number_of_nodes(), '| Friendships:', G.number_of_edges())
```

```python
measures = {'Degree (popular)': nx.degree_centrality(G), 'Closeness (spread)': nx.closeness_centrality(G), 'Betweenness (bridge)': nx.betweenness_centrality(G, k=min(200, len(G)), seed=42)}
for name, cent in measures.items():
    top = sorted(cent.items(), key=lambda x: -x[1])[:5]
    print(f'\n{name}: ' + ', '.join(f'User {u} = {s:.4f}' for u, s in top))
```

```python
deg = nx.degree_centrality(G)
nx.draw(G, nx.spring_layout(G, seed=42, k=0.15), node_size=[20 + 5000 * deg[n] for n in G.nodes()], node_color=list(deg.values()), cmap='viridis', edge_color='gray', alpha=0.7)
plt.title('Social Network Analysis'); plt.axis('off'); plt.show()
```
